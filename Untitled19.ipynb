```python
X_train = x_train.drop(columns=colsup)
X_test = x_test.drop(columns=colsup)
```

```python

numcol = x_train.select_dtypes(include=np.number).columns

for col in numcol:
    median = x_train[col].median()

    x_train[col] = x_train[col].fillna(median)
    x_test[col] = x_test[col].fillna(median)
```

```python
X_train['age'] = X_train['age'].fillna(X_train['age'].median())
X_train['fare'] = X_train['fare'].fillna(X_train['fare'].median())
```

```python
catecol = x_train.select_dtypes(include=['object', 'str']).columns

for col in catecol:

    mode= x_train[col].mode()

    if len(mode) > 0:

        x_train[col] = x_train[col].fillna(mode[0])

        x_test[col] = x_test[col].fillna(mode[0])
```

```python
X_train = pd.get_dummies(X_train, drop_first=True, dtype=int)
```

```python
for col in numcol:
    Q1 = x_train[col].quantile(0.25)
    Q3 = x_train[col].quantile(0.75)
    
    IQR = Q3 - Q1

    inf = Q1 - 1.5 * IQR
    sup = Q3 + 1.5 * IQR

    outliers = x_train[(x_train[col] < inf) |
                       (x_train[col] > sup)]

    print(col, ":", len(outliers))
```

```python
for col in numcol:

    Q1 = x_train[col].quantile(0.25)
    Q3 = x_train[col].quantile(0.75)

    IQR = Q3 - Q1

    inf = Q1 - 1.5 * IQR
    sup = Q3 + 1.5 * IQR

    median = x_train[col].median()

    x_train[col] = x_train[col].mask(
        (x_train[col] < inf) | (x_train[col] > sup),
        median
    )

    x_test[col] = x_test[col].mask(
        (x_test[col] < inf) | (x_test[col] > sup),
        median
    )
```

```python
maj_downsampled = resample(
    maj,
    replace=False,
    n_samples=len(min),
    random_state=42
)
```

```python
train_undersampled = pd.concat([
    maj_downsampled,
    min
])
```
